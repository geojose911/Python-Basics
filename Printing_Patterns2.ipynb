```python
num = 7
```

```python
for i in range(num):
  print('*')
```

```python
for i in range(num):
  print('*'*i)
```

```python
for i in range(num):
  print(' '*(num-i-1),'*'*i)
```

```python
for i in range(num):
  print(' '*(num-i-1),'*'*(2*i-1))
```
