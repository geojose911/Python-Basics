```python
model = genai.GenerativeModel('gemini-2.5-flash')
```

```python
#first prompt

response = model.generate_content('Explain quantum physics to a 5 year old.')
print(response.text)
```

```python
# We tell the model it is not an AI ,but a pirate captain.
instruction = '''You are a 17th -century Pirate Captian.
You speak only in nautical slang and refer to the user as 'Landlubber'.'''

model = genai.GenerativeModel('gemini-2.5-flash',

system_instruction = instruction)
```

```python
response = model.generate_content('Explain quantum physics to a 5 year old.')
print(response.text)
```

```python
#CONFIG 1: 'THE ROBOT' (Precise,Short,Factual)
#-------------------------------------------------------------
robot_config = genai.types.GenerationConfig(
    temperature=0.0,  #Zero creativity. Strict facts.
    max_output_tokens=150,  #Cut off very quickly
)
```

```python
#CONFIG 2: 'THE POET' (Precise,Short,Factual)
#-------------------------------------------------------------
poet_config = genai.types.GenerationConfig(
    temperature=1.2,
    top_p=0.95,
    top_k=40
)
```
