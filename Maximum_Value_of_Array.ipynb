Maximum Value of Array of N numbers

```python
arr = [12, 45, 2, 89, 34, 67, 10]
num = arr[0]
for i in arr:
  if i>num:
    num=i
print(num)
```

Other Methords

```python
n = int(input("Enter number of elements (N): "))
numbers = list(map(int, input(f"Enter {n} numbers separated by space: ").split()))
print("Maximum Value:", max(numbers))
```

```python
arr = [12, 45, 2, 89, 34, 67, 10]
max_value = arr[0]
for num in arr:
    if num > max_value:
        max_value = num
        print("Maximum value:", max_value)
```
