```python
#if - elif - else condition

# biggest of 3 numbers

a = float(input ('enter 1st: '))
b = float(input('enter 2nd: '))
c = float(input("enter 3rd: "))
if a>b and a>c:
  print(a,' is biggest')
elif a<b and b>c:
  print(b,' is biggest')
else:
  print(c,' is biggest')
```
