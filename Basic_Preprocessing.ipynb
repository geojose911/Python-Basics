Detecting Outliers

```python
plt.boxplot(data['beer_servings'])
plt.title('Boxplot of beer_servings')
plt.show()
```

```python
plt.boxplot(data['spirit_servings'])
plt.title('Boxplot of spirit_servings')
plt.show()
```

```python
Q1 = np.percentile(data['spirit_servings'],25)
Q2 = np.percentile(data['spirit_servings'],50)
Q3 = np.percentile(data['spirit_servings'],75)
print(Q1)
print(Q2)
print(Q3)
```

```python
data['spirit_servings'].median()
```

```python
IQR = Q3-Q1
print(IQR)
```

```python
up_lim = Q3+1.5*IQR
low_lim = Q1-1.5*IQR
```

```python
print(up_lim)
print(low_lim)
```

```python
outlier = []
for x in data['spirit_servings']:
    if x > up_lim or x < low_lim:
        outlier.append(x)
```
