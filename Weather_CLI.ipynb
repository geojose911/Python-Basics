```python
import requests

def get_weather(city):
    url = f"https://wttr.in/{city}?format=%C+%t+%h+%w"
    response = requests.get(url)
    print(f"Weather in {city}: {response.text}")

get_weather("Kochi")
get_weather("London")
```
